# Text Classification with TF-IDF
This notebook creates source data in CSV format, transforms text into TF-IDF features, trains a classifier, and evaluates model accuracy.

```python
import pandas as pd

rows = [
    ("battery level low warning", "Battery"),
    ("battery critically low", "Battery"),
    ("device battery needs replacement", "Battery"),
    ("power supply unstable", "Battery"),
    ("motion detected in living room", "Security"),
    ("motion detected at main gate", "Security"),
    ("unauthorized access attempt detected", "Security"),
    ("intrusion detected in warehouse", "Security"),
    ("temperature above threshold", "Temperature"),
    ("temperature sensor reading high", "Temperature"),
    ("overheating detected in motor", "Temperature"),
    ("cooling system failure temperature rising", "Temperature"),
    ("sensor failure detected", "Fault"),
    ("pressure sensor malfunction", "Fault"),
    ("device not responding", "Fault"),
    ("communication error detected", "Fault"),
    ("system running normally", "Normal"),
    ("device operating within limits", "Normal"),
    ("all sensors functioning properly", "Normal"),
    ("routine status update normal", "Normal"),
    ("gas leakage detected alert", "Alert"),
    ("smoke detected in kitchen", "Alert"),
    ("fire alarm triggered", "Alert"),
    ("water leakage detected in basement", "Alert")
]

df = pd.DataFrame(rows, columns=["Message", "Category"])
csv_path = "data_log.csv"
df.to_csv(csv_path, index=False)
print(f"Saved source data to: {csv_path}")
df.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

data = pd.read_csv("data_log.csv")
X = data["Message"]
y = data["Category"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("TF-IDF training matrix shape:", X_train_tfidf.shape)
print("TF-IDF test matrix shape:", X_test_tfidf.shape)
```

```python
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

y_pred = model.predict(X_test_tfidf)
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```
